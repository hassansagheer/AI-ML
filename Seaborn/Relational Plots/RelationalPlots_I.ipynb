```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")
tips.head()
```

Basic Relational Plot

```python
sns.set(style="ticks")

sns.relplot(
    x="total_bill",
    y="tip",
    data=tips
)

plt.show()
```

Group Data Using Hue

```python
sns.relplot(
    x="total_bill",
    y="tip",
    hue="time",
    data=tips
)

plt.show()
```

Created Faceted Relational Plots

```python
sns.relplot(
    x="total_bill",
    y="tip",
    hue="day",
    col="time",
    row="sex",
    data=tips
)

plt.show()
```

Use Different Point Sizes

```python
sns.relplot(
    x="total_bill",
    y="tip",
    hue="day",
    size="size",
    data=tips
)

plt.show()
```

Create Line Relational Plot

```python
sns.relplot(
    x="total_bill",
    y="tip",
    kind="line",
    data=tips
)

plt.show()
```
