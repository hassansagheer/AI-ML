```python
# import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# set grid style
sns.set(style ="darkgrid")

# import dataset
dataset = pd.read_csv('FuelConsumption.csv')
```

```python
sns.relplot(x ="ENGINESIZE", y ="CO2EMISSIONS",
            data = dataset);
```

```python
sns.relplot(x ="ENGINESIZE", y ="CO2EMISSIONS", 
            hue ="FUELTYPE", data = dataset);
```

```python
sns.relplot(x ="ENGINESIZE", y ="CO2EMISSIONS",
            hue ="FUELTYPE", style ="FUELTYPE",
            data = dataset);
```

```python
sns.relplot(x ="ENGINESIZE", y ="CO2EMISSIONS",
            hue ="CYLINDERS", data = dataset);
```

```python
sns.relplot(x ="ENGINESIZE", y ="CO2EMISSIONS", 
            size ="CYLINDERS", data = dataset);
```

```python
sns.relplot(x ="ENGINESIZE", y ="CO2EMISSIONS", 
            hue="FUELCONSUMPTION_CITY", style="FUELTYPE", data = dataset);
```

```python
sns.relplot(x ="ENGINESIZE", y ="CO2EMISSIONS", 
            hue="FUELCONSUMPTION_CITY", size="FUELCONSUMPTION_HWY", data = dataset);
```
