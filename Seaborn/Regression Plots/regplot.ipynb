```python
# importing required packages
import seaborn as sns
import matplotlib.pyplot as plt

# loading dataset
data = sns.load_dataset("mpg")

# draw regplot
sns.regplot(x = "mpg", 
            y = "acceleration", 
            data = data)

# show the plot
plt.show()
```

```python
# importing required packages
import seaborn as sns
import matplotlib.pyplot as plt

# loading dataset
data = sns.load_dataset("titanic")

# draw regplot
sns.regplot(x = "age",
            y = "fare",
            data = data,
            dropna = True)
# show the plot
plt.show()
```

```python
# importing required packages
import seaborn as sns

# loading dataset
data = sns.load_dataset("mpg")
sns.regplot(
    data=data, x="weight", y="horsepower",
    ci=99, marker="x", color=".3", line_kws=dict(color="r"),
)
```

```python
# importing required packages
import seaborn as sns
import numpy as np

# loading dataset
data = sns.load_dataset("mpg")
sns.regplot(data=data, x="weight", y="mpg", x_bins=np.arange(2000, 5500, 250), order=2)
```

```python
# importing required packages
import seaborn as sns
import numpy as np

# loading dataset
data = sns.load_dataset("mpg")
sns.regplot(data=data, x="cylinders", y="acceleration", x_estimator=np.mean, order=2)
```
