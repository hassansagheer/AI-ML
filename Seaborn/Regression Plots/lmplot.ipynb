```python
import seaborn as sns
import matplotlib.pyplot as plt

# loading dataset
df = sns.load_dataset('tips')

# scatter plot with regression 
# line(by default)
sns.lmplot(x ='total_bill', y ='tip', data = df)

# Show the plot
plt.show()
```

```python
# scatter plot without regression
# line.
sns.lmplot(x ='total_bill', y ='tip', fit_reg = False, data = df)
```

```python
sns.lmplot(x ='total_bill', y ='tip', fit_reg = False, hue = 'sex', data = df)
```

```python
import seaborn as sns

# loading dataset
data = sns.load_dataset("penguins")
sns.lmplot(data=data, x="bill_length_mm", y="bill_depth_mm", hue="species")
```

```python
import seaborn as sns

# loading dataset
data = sns.load_dataset("penguins")
sns.lmplot(
    data=data, x="bill_length_mm", y="bill_depth_mm",
    hue="species", col="sex", height=4,
)
```

```python
import seaborn as sns

# loading dataset
data = sns.load_dataset("penguins")
sns.lmplot(
    data=data, x="bill_length_mm", y="bill_depth_mm",
    col="species", row="sex", height=3,
)
```

```python
import seaborn as sns

# loading dataset
data = sns.load_dataset("penguins")
sns.lmplot(
    data=data, x="bill_length_mm", y="bill_depth_mm",
    col="species", row="sex", height=3,
    facet_kws=dict(sharex=False, sharey=False),
)
```
