```python
import matplotlib.pyplot as plt
import seaborn

# Set style
seaborn.set(style="whitegrid")
tips = seaborn.load_dataset("tips")

# Use catplot with 'col="sex"' to create separate columns/panels for each sex
g = seaborn.catplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="smoker",
    col="sex",          # Splits the graph by sex (Male vs Female)
    kind="swarm",
    palette="Set2",
    dodge=True,
    height=5,
    aspect=0.8,
)

# Set labels and title for the figure grid
g.set_axis_labels("Days", "Total Bill ($)")
g.fig.subplots_adjust(top=0.8)
g.fig.suptitle('Smoker Distribution Split by Sex')

# Move the legend outside the plot to the top right
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', borderaxespad=0, title='Smoker')

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn

# Set style
seaborn.set(style="whitegrid")
tips = seaborn.load_dataset("tips")

# Use catplot with 'col="sex"' to create separate columns/panels for each sex
g = seaborn.catplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="smoker",
    col="sex",          # Splits the graph by sex (Male vs Female)
    kind="swarm",
    palette="Set2",
    dodge=True,
    height=5,
    aspect=0.8,
)

# Set labels and title for the figure grid
g.set_axis_labels("Days", "Total Bill ($)")
g.fig.subplots_adjust(top=0.8)
g.fig.suptitle('Smoker Distribution Split by Sex')

# Move catplot's single built-in legend outside the plot to the top right
g._legend.set_bbox_to_anchor((1.05, 1))
g._legend.set_loc('upper left')
g._legend.set_title('Smoker')

plt.tight_layout()
plt.show()
```
