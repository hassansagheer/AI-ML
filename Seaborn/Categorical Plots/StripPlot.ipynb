```python
import seaborn
import matplotlib.pyplot as plt
     
seaborn.set(style = 'whitegrid')  
tip = seaborn.load_dataset("tips")  
     
# Instead of calling plt.xlabel() and plt.ylabel() separately:
plt.gca().set(xlabel='Days', ylabel='Total Bill ($)', title='Tips Distribution')

seaborn.stripplot(x="day", y="total_bill", data=tip, hue="smoker", palette="viridis")

# Place legend outside the graph on the top right
plt.legend(title='Smoker',bbox_to_anchor=(1.02, 1), loc='upper left', borderaxespad=0)

# Adjust layout to prevent the legend from getting cut off
plt.tight_layout()

plt.show()
```

```python
import seaborn
import matplotlib.pyplot as plt
     
seaborn.set(style = 'whitegrid')  
tip = seaborn.load_dataset("tips")  
     
# Instead of calling plt.xlabel() and plt.ylabel() separately:
plt.gca().set(xlabel='Days', ylabel='Total Bill ($)', title='Tips Distribution')

seaborn.stripplot(x="day", y="total_bill", data=tip, hue="smoker", palette="Set1", size=20, marker="s", alpha=0.2)

# Place legend outside the graph on the top right
plt.legend(title='Smoker',bbox_to_anchor=(1.02, 1), loc='upper left', borderaxespad=0)

# Adjust layout to prevent the legend from getting cut off
plt.tight_layout()

plt.show()
```
