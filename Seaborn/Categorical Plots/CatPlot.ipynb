```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Read the tips dataset
tips = sns.load_dataset("tips")

# Use catplot with kind="point"
sns.catplot(x='size', y='tip', data=tips, hue='sex', kind='bar', errorbar=None)

# Show the plot
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Read the tips dataset
tips = sns.load_dataset("tips")

# Use catplot with kind="point"
sns.catplot(x='size', y='tip', data=tips, hue='sex', palette='viridis', kind='swarm', aspect=4)

# Show the plot
plt.show()
```
