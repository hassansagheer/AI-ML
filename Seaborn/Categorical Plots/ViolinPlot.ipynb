```python
import seaborn
     
seaborn.set(style = 'whitegrid') 
fmri = seaborn.load_dataset("fmri") 
   
seaborn.violinplot(x ="timepoint", 
             y ="signal", 
             data = fmri)
```

```python
import seaborn as sns
   
sns.set(style = 'whitegrid') 
fmri = sns.load_dataset("fmri") 
   
sns.violinplot(x ="timepoint", 
             y ="signal", 
             hue ="region", 
             data = fmri)
```
