```python
# importing the required module
import seaborn
import matplotlib.pyplot as plt
# use to set style of background of plot
seaborn.set(style="whitegrid")

# loading data-set
tips = seaborn.load_dataset("tips")

# Instead of calling plt.xlabel() and plt.ylabel() separately:
plt.gca().set(xlabel='Days', ylabel='Total Bill ($)', title='Tips Distribution')

seaborn.violinplot(x="day", y="total_bill", hue="time", dodge=False, data=tips)
```
