```python
import matplotlib.pyplot as plt
import seaborn

seaborn.set(style='whitegrid')
fmri = seaborn.load_dataset("fmri")

# Modify size of marker (s) and color palette (palette)
seaborn.swarmplot(
    x="timepoint",
    y="signal",
    hue="timepoint",      # Maps each timepoint to a unique color
    palette="rainbow",    # Applies rainbow palette
    s=4,                  # Adjust marker size (s or size, e.g., 4 or 5 to prevent heavy overlapping)
    legend=False,         # Hides duplicate legend
    data=fmri
)

plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn

# use to set style of background of plot
seaborn.set(style="whitegrid")

# loading data-set
tips = seaborn.load_dataset("tips")

# Create the swarmplot
seaborn.swarmplot(x="total_bill", y="day", data=tips, hue="tip", palette="viridis", dodge=False)

# Move the legend outside the plot to the top right
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', borderaxespad=0.)

# Adjust layout to ensure the legend is not cut off
plt.tight_layout()

# Show the plot
plt.show()
```

```python
# importing the required module
import time
import seaborn
import matplotlib.pyplot as plt

# use to set style of background of plot
seaborn.set(style="whitegrid")

# loading data-set
tips = seaborn.load_dataset("tips")

# Instead of calling plt.xlabel() and plt.ylabel() separately:
plt.gca().set(xlabel='Days', ylabel='Total Bill ($)', title='Time Distribution')

seaborn.swarmplot(x="day", y="total_bill", data=tips, hue="time", palette="RdBu", edgecolor="grey", linewidth=2)

# Move the legend outside the plot to the top right
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', borderaxespad=0)

# Adjust layout to ensure the legend is not cut off
plt.tight_layout()

# Show the plot
plt.show()
```

```python
# importing the required module
import seaborn

# use to set style of background of plot
seaborn.set(style="whitegrid")

# loading data-set
tips = seaborn.load_dataset("tips")

# Instead of calling plt.xlabel() and plt.ylabel() separately:
plt.gca().set(xlabel='Days', ylabel='Total Bill ($)', title='Smoker Distribution')

seaborn.swarmplot(x="day", y="total_bill", hue="smoker", data=tips, palette="Set2", dodge=True)

# Move the legend outside the plot to the top right
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', borderaxespad=0, title='Smoker')

# Adjust layout to ensure the legend is not cut off
plt.tight_layout()

# Show the plot
plt.show()
```

```python
# importing the required module
import seaborn

# use to set style of background of plot
seaborn.set(style="whitegrid")

# loading data-set
tips = seaborn.load_dataset("tips")

seaborn.swarmplot(x="day", y="total_bill", hue="smoker",
                   data=tips, palette="Set2", size=20, marker="D",
                   edgecolor="gray", alpha=.25)

# Move the legend outside the plot to the top right
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', borderaxespad=0, title='Smoker')

# Adjust layout to ensure the legend is not cut off
plt.tight_layout()

# Show the plot
plt.show()
```
