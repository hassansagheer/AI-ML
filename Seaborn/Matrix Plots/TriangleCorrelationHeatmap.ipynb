```python
# import modules
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# import file with data
data = sns.load_dataset("glue").pivot(index="Model", columns="Task", values="Score")

# creating mask
mask = np.triu(np.ones_like(data.corr()))

# plotting a triangle correlation heatmap
sns.heatmap(data.corr(), cmap="crest", annot=True, linewidth=.5, mask=mask)

# displaying heatmap
plt.show()
```
