```python
import numpy as np
m = np.matrix([[1,-2,3],[0,4,5],[7,8,-9]]);
print('Matrix A: \n',m);
```

```python
print('Transpose of Matrix A: \n',m.T);
```

```python
print('Inverse of Matrix A: \n',m.I);
```

```python
v = np.matrix([[2],[3],[4]]);
print('Matrix B: \n',v);
```

```python
# product of two matrices
print('Product of Matrix A x B\n', m*v);
```

```python
# Determinent of Matrix A i.e. m
import numpy.linalg
numpy.linalg.det(m)
```

```python
numpy.linalg.eigvals(m)
```

```python
# Find x matrix m*x=v
x = numpy.linalg.solve(m,v)
print('Matrix X: \n',x);
```
