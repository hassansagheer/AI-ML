```python
import numpy as np
from scipy import signal

sig = np.repeat([0., 1., 0.], 100)
win = signal.windows.hann(50)

# 1D Convolution (normalized by window sum)
filtered = signal.convolve(sig, win, mode='same') / sum(win)

# FFT-based Convolution (much faster for large arrays)
fft_filtered = signal.fftconvolve(sig, win, mode='same')

# Cross-correlation (finding time lag between two signals)
x = np.sin(np.linspace(0, 10, 100))
y = np.roll(x, 15)  # Shift signal by 15 samples
corr = signal.correlate(y, x, mode='full')
lags = signal.correlation_lags(len(y), len(x), mode='full')
best_lag = lags[np.argmax(corr)]  # Returns 15
```
