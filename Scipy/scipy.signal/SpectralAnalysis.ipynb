```python
import numpy as np
from scipy import signal

fs = 1000.0
t = np.linspace(0, 5, int(5 * fs), endpoint=False)
x = np.sin(2 * np.pi * 120 * t) + np.random.normal(0, 1, size=t.shape)

# Welch's method (smooth Power Spectral Density estimate)
freqs, psd = signal.welch(x, fs=fs, nperseg=1024)

# Periodogram (raw unaveraged PSD)
f_per, Pxx = signal.periodogram(x, fs=fs)

# Short-Time Fourier Transform (STFT) & Spectrogram
f_stft, t_stft, Zxx = signal.stft(x, fs=fs, nperseg=256)
f_spec, t_spec, Sxx = signal.spectrogram(x, fs=fs, nperseg=256)

# Lomb-Scargle Periodogram (for unevenly sampled time series)
t_uneven = np.sort(np.random.uniform(0, 10, 500))
y_uneven = np.sin(2.5 * t_uneven)
ang_freqs = np.linspace(0.1, 10, 1000)
pgram = signal.lombscargle(t_uneven, y_uneven, ang_freqs, normalize=True)
```
