```python
import numpy as np
from scipy import signal

fs = 1000.0
sos = signal.butter(N=4, Wn=100, btype='low', fs=fs, output='sos')

# Frequency response for SOS filter
freqs, h = signal.sosfreqz(sos, worN=1024, fs=fs)
magnitude_db = 20 * np.log10(np.maximum(np.abs(h), 1e-10))
phase_rad = np.unwrap(np.angle(h))

# Frequency response for Transfer Function (b, a) filter
b, a = signal.butter(4, 100, fs=fs)
w, h_ba = signal.freqz(b, a, fs=fs)
```
