```python
import numpy as np
from scipy import signal

t = np.linspace(0, 1, 1000, endpoint=False)
x = np.sin(2 * np.pi * 10 * t)

# Fourier-based resampling (exact for periodic signals, resamples to 250 points)
x_resampled = signal.resample(x, num=250)

# Polyphase resampling (efficient, avoids edge ringing; upsample by 3, downsample by 2)
x_poly = signal.resample_poly(x, up=3, down=2)

# Decimation (applies anti-aliasing lowpass filter before downsampling by factor q)
x_decimated = signal.decimate(x, q=4, ftype='iir', zero_phase=True)
```
