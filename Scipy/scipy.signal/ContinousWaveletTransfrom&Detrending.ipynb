```python
import numpy as np
from scipy import signal

# Remove linear trend or constant offset (DC component)
t = np.linspace(0, 10, 200)
drifting_sig = np.sin(2 * np.pi * t) + 0.5 * t + 3.0
detrended_linear = signal.detrend(drifting_sig, type='linear')
detrended_dc = signal.detrend(drifting_sig, type='constant')

# Ricker (Mexican hat) wavelet generation
wavelet = signal.ricker(points=100, a=4.0)
```
