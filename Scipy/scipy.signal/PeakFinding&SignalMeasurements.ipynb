```python
import numpy as np
from scipy import signal

x = np.sin(np.linspace(0, 6 * np.pi, 500)) + 0.2 * np.random.randn(500)

# Find peaks with constraints on height, minimum distance, and prominence
peaks, properties = signal.find_peaks(x, height=0.5, distance=50, prominence=0.8)

# Measure peak widths at half-maximum (rel_height=0.5)
widths, width_heights, left_ips, right_ips = signal.peak_widths(x, peaks, rel_height=0.5)

# Analytic signal via Hilbert transform (extract instantaneous amplitude & phase)
analytic_signal = signal.hilbert(x)
amplitude_envelope = np.abs(analytic_signal)
instantaneous_phase = np.unwrap(np.angle(analytic_signal))
```
