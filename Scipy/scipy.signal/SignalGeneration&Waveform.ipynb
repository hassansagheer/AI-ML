```python
import numpy as np
from scipy import signal

t = np.linspace(0, 1, 500, endpoint=False)

# Square wave (5 Hz, 50% duty cycle)
sq = signal.square(2 * np.pi * 5 * t, duty=0.5)

# Sawtooth / Triangle wave (width=0.5 makes it a triangle wave)
saw = signal.sawtooth(2 * np.pi * 5 * t, width=1.0)

# Chirp (frequency sweep from 6 Hz to 1 Hz over 1 second)
chirp_sig = signal.chirp(t, f0=6, f1=1, t1=1, method='linear')

# Gaussian-modulated sinusoidal pulse
t_gaus = np.linspace(-1, 1, 200, endpoint=False)
gaus_pulse = signal.gausspulse(t_gaus, fc=5, bw=0.5)

# Unit impulse (delta function)
imp = signal.unit_impulse(100, idx='mid')
```
