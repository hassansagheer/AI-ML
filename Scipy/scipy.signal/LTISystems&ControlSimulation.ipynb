```python
import numpy as np
from scipy import signal

# Define continuous transfer function: H(s) = 1 / (s^2 + 2s + 1)
num = [1.0]
den = [1.0, 2.0, 1.0]
sys = signal.TransferFunction(num, den)

# Step response
t_step, y_step = signal.step(sys)

# Impulse response
t_imp, y_imp = signal.impulse(sys)

# Simulate response to an arbitrary input signal u
t = np.linspace(0, 10, 500)
u = np.sin(t)
t_out, y_out, x_state = signal.lsim(sys, U=u, T=t)

# Bode plot data (frequency, magnitude in dB, phase in degrees)
w, mag, phase = signal.bode(sys)
```
