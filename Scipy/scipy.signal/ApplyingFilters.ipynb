```python
import numpy as np
from scipy import signal

fs = 1000.0
t = np.linspace(0, 1, int(fs), endpoint=False)
noisy_sig = np.sin(2 * np.pi * 10 * t) + 0.5 * np.sin(2 * np.pi * 250 * t)

# Design filter
sos = signal.butter(N=4, Wn=50, btype='low', fs=fs, output='sos')

# Zero-phase filtering (applies filter forward and backward; eliminates phase shift)
clean_zero_phase = signal.sosfiltfilt(sos, noisy_sig)

# Causal filtering (real-time simulation; introduces phase delay)
clean_causal = signal.sosfilt(sos, noisy_sig)

# Applying an FIR filter (numerator b = fir_taps, denominator a = 1.0)
fir_taps = signal.firwin(51, 50, fs=fs)
clean_fir = signal.filtfilt(fir_taps, [1.0], noisy_sig)
```
