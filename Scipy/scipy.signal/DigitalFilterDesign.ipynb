```python
import numpy as np
from scipy import signal

fs = 1000.0  # Sampling frequency (Hz)

# --- IIR Filters (Infinite Impulse Response) ---
# Butterworth Bandpass (50 Hz to 200 Hz)
sos_butter = signal.butter(N=4, Wn=[50, 200], btype='bandpass', fs=fs, output='sos')

# Chebyshev Type I Lowpass (cutoff 100 Hz, 1 dB passband ripple)
sos_cheby1 = signal.cheby1(N=4, rp=1, Wn=100, btype='low', fs=fs, output='sos')

# Elliptic Highpass (cutoff 150 Hz, 1 dB ripple, 40 dB stopband attenuation)
sos_ellip = signal.ellip(N=4, rp=1, rs=40, Wn=150, btype='high', fs=fs, output='sos')

# IIR Notch Filter (remove 60 Hz powerline hum)
b_notch, a_notch = signal.iirnotch(w0=60.0, Q=30.0, fs=fs)

# --- FIR Filters (Finite Impulse Response) ---
# Window method (101 taps, lowpass at 100 Hz)
fir_taps = signal.firwin(numtaps=101, cutoff=100, window='hamming', fs=fs)

# Parks-McClellan (Remez exchange) optimal equiripple filter
remez_taps = signal.remez(numtaps=101, bands=[0, 100, 120, 500], desired=[1, 0], fs=fs)
```
