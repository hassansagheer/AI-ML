```python
import numpy as np
from scipy.fft import fft, fftfreq, fftshift, ifftshift

fs = 8.0
freqs = fftfreq(8, 1 / fs)
spectrum = np.array([10, 2, 1, 0, 0, 0, 1, 2])  # Simulated FFT magnitudes

shifted_freqs = fftshift(freqs)
shifted_spec = fftshift(spectrum)

print("Unshifted freqs: ", freqs)
print("Shifted freqs:   ", shifted_freqs)
print("Shifted spectrum:", shifted_spec)
print("Restored spectrum:", ifftshift(shifted_spec))
```
