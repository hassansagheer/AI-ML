```python
import numpy as np
from scipy.fft import dst, idst

data = np.array([1.0, 2.0, 3.0, 2.0, 1.0])

transformed = dst(data, type=2, norm="ortho")
restored = idst(transformed, type=2, norm="ortho")

print("DST Coefficients:", np.round(transformed, 3))
print("Restored Data:   ", np.round(restored, 3))
```
