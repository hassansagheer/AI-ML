```python
import numpy as np
from scipy.fft import irfft, rfft, rfftfreq

rng = np.random.default_rng(42)
fs = 200.0
t = np.linspace(0, 1, int(fs), endpoint=False)

# Clean 5 Hz signal + random white noise
clean = np.sin(2 * np.pi * 5 * t)
noisy = clean + 0.8 * rng.standard_normal(len(t))

# Transform to frequency domain
yf = rfft(noisy)
magnitudes = np.abs(yf)

# Threshold filter: keep only frequencies with magnitude > 30
yf_filtered = np.where(magnitudes > 30, yf, 0)
denoised = irfft(yf_filtered)

print("Noisy MSE vs Clean:   ", round(np.mean((noisy - clean) ** 2), 4))
print("Denoised MSE vs Clean:", round(np.mean((denoised - clean) ** 2), 4))
```
