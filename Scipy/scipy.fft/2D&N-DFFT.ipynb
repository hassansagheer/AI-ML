```python
import numpy as np
from scipy.fft import fft2, fftn, ifft2, ifftn

# 2D array (e.g., grayscale image patch)
image = np.array([
    [1, 2, 1],
    [2, 4, 2],
    [1, 2, 1],
])

f2 = fft2(image)
inv2 = ifft2(f2).real

# 3D array (N-dimensional)
volume = np.ones((2, 2, 2))
fn = fftn(volume)

print("2D FFT DC Component (top-left):", f2[0, 0])
print("2D Reconstructed:\n", np.round(inv2, 1))
print("3D FFT Sum (DC Component):     ", fn[0, 0, 0])
```
