```python
import numpy as np
from scipy.fft import dct, idct

data = np.array([4.0, 3.0, 5.0, 10.0, 5.0, 3.0])

# Type-II DCT with orthonormal scaling
coefficients = dct(data, type=2, norm="ortho")

# Zero out high-frequency components (lossy compression simulation)
compressed_coeffs = coefficients.copy()
compressed_coeffs[3:] = 0

reconstructed = idct(compressed_coeffs, type=2, norm="ortho")

print("DCT Coefficients: ", np.round(coefficients, 2))
print("Compressed Reconst:", np.round(reconstructed, 2))
```
