```python
import numpy as np
from scipy.fft import irfft, rfft, rfftfreq

# 100 Hz sampling rate, 1 second duration, mixed 10 Hz + 25 Hz signal
fs = 100.0
t = np.linspace(0, 1, int(fs), endpoint=False)
signal = 3.0 * np.sin(2 * np.pi * 10 * t) + 1.5 * np.cos(2 * np.pi * 25 * t)

# Compute Real FFT
yf = rfft(signal)
xf = rfftfreq(len(signal), 1 / fs)

# Find dominant frequencies (amplitude = 2 * |yf| / N)
amplitudes = 2.0 / len(signal) * np.abs(yf)
peaks = xf[amplitudes > 0.5]

print("Output array length:", len(yf))
print("Detected peak frequencies (Hz):", peaks)
print("Matches original signal:", np.allclose(signal, irfft(yf)))
```
