```python
import numpy as np
from scipy.fft import fft, fftfreq, ifft

# Generate a 1-second signal sampled at 8 Hz (5 Hz sine wave)
fs = 8.0
t = np.arange(0, 1, 1 / fs)
signal = np.sin(2 * np.pi * 2 * t)  # 2 Hz wave

# Compute FFT and corresponding frequency bins
yf = fft(signal)
xf = fftfreq(len(signal), 1 / fs)

# Reconstruct original signal
reconstructed = ifft(yf)

print("Frequencies (Hz):", xf)
print("FFT Magnitudes:  ", np.round(np.abs(yf), 2))
print("Reconstructed:   ", np.round(reconstructed.real, 2))
```
