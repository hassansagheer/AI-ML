```python
import numpy as np
from scipy import sparse
from scipy.sparse.linalg import spsolve

# 1. Special builders
I = sparse.eye_array(3, format='csr')
T = sparse.diags_array([-1, 2, -1], offsets=[-1, 0, 1], shape=(3, 3), format='csr')

# 2. Stacking matrices horizontally / vertically
H = sparse.hstack([T, I], format='csr')

# 3. Solving sparse linear system Tx = b
b = np.array([1.0, 0.0, 1.0])
x = spsolve(T, b)

print("Tridiagonal Matrix T:\n", T.toarray())
print("\nHorizontal Stack [T | I]:\n", H.toarray())
print("\nSolution x to Tx = b:", x)
```
