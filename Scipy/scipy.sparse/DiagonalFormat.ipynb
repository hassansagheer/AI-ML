```python
import numpy as np
from scipy.sparse import dia_array

# 3 diagonals of length 4
data = np.array([
    [1, 2, 3, 4],    # Main diagonal (offset 0)
    [0, 5, 6, 7],    # Upper diagonal (offset +1; first element ignored)
    [8, 9, 10, 0]    # Lower diagonal (offset -1; last element ignored)
])
offsets = np.array([0, 1, -1])

dia = dia_array((data, offsets), shape=(4, 4))

print("DIA dense output:\n", dia.toarray())
```
