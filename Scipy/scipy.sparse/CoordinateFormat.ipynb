```python
import numpy as np
from scipy.sparse import coo_array

rows = np.array([0, 0, 1, 2, 2, 0])  # Note duplicate (0, 0)
cols = np.array([0, 3, 2, 0, 2, 0])
vals = np.array([1, 2, 3, 4, 5, 10])

coo = coo_array((vals, (rows, cols)), shape=(3, 4))

print("COO Representation:")
print(coo)
print("\nConverted to CSR (duplicates summed):\n", coo.tocsr().toarray())
```
