```python
from scipy.sparse import lil_array

lil = lil_array((3, 4), dtype=int)

# Assign values incrementally
lil[0, 0] = 1
lil[0, 3] = 2
lil[1, 2] = 3
lil[2, [0, 2]] = [4, 5]

print("LIL internal rows:", lil.rows)
print("LIL internal data:", lil.data)
print("\nDense output:\n", lil.toarray())
```
