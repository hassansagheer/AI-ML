```python
import numpy as np
from scipy.sparse import csr_array

# Create from a dense 2D array
dense = np.array([
    [1, 0, 0, 2],
    [0, 0, 3, 0],
    [4, 0, 5, 0]
])
csr = csr_array(dense)

print("CSR Internal Arrays:")
print("Data:   ", csr.data)     # Non-zero values
print("Indices:", csr.indices)  # Column indices of values
print("Indptr: ", csr.indptr)   # Row start/end pointers
print("\nRow 0 slice:", csr[[0], :].toarray())
print("Matrix-Vector product:", csr @ np.array([1, 2, 3, 4]))
```
