```python
import numpy as np
from scipy.sparse import csc_array

dense = np.array([
    [1, 0, 0, 2],
    [0, 0, 3, 0],
    [4, 0, 5, 0]
])
csc = csc_array(dense)

print("CSC Internal Arrays:")
print("Data:   ", csc.data)     # Non-zero values (column-major order)
print("Indices:", csc.indices)  # Row indices of values
print("Indptr: ", csc.indptr)   # Column start/end pointers
print("\nColumn 2 slice:\n", csc[:, [2]].toarray())
```
