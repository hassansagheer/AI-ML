```python
import numpy as np
from scipy.sparse import bsr_array

# Two 2x2 dense blocks
blocks = np.array([
    [[1, 2], [3, 4]],  # Block 0
    [[5, 6], [7, 8]]   # Block 1
])
indices = np.array([0, 2])  # Block column indices
indptr = np.array([0, 1, 2]) # Block row pointers

# Shape is (2 block-rows * 2, 3 block-cols * 2) = (4, 6)
bsr = bsr_array((blocks, indices, indptr), shape=(4, 6))

print("Block size:", bsr.blocksize)
print("BSR dense output:\n", bsr.toarray())
```
