```python
from scipy.sparse import dok_array

dok = dok_array((3, 4), dtype=float)

# Dictionary-style assignment and lookup
dok[0, 1] = 2.5
dok[2, 3] = 7.1

print("Keys in DOK:", list(dok.keys()))
print("Value at (2, 3):", dok[2, 3])
print("Value at unset (1, 1):", dok[1, 1])
print("\nDense output:\n", dok.toarray())
```
