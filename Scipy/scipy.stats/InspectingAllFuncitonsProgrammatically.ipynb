```python
from scipy import stats

# List all continuous distributions
continuous_dists = [
    d for d in dir(stats) if isinstance(getattr(stats, d), stats.rv_continuous)
]

# List all discrete distributions
discrete_dists = [
    d for d in dir(stats) if isinstance(getattr(stats, d), stats.rv_discrete)
]

print(f"Continuous ({len(continuous_dists)}):", continuous_dists)
print(f"Discrete ({len(discrete_dists)}):", discrete_dists)
```
