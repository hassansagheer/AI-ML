```python
import numpy as np
from scipy import stats

# 1. Chi-Square Goodness of Fit (observed vs expected frequencies)
observed = [50, 30, 20]
expected = [40, 40, 20]
chi2_stat, p_val = stats.chisquare(f_obs=observed, f_exp=expected)

# 2. Chi-Square Test of Independence (2D contingency table)
table = np.array([[30, 10], [15, 25]])
chi2_stat, p_val, dof, ex_freq = stats.chi2_contingency(table)

# 3. Fisher's Exact Test (for 2x2 contingency tables with small sample sizes)
odds_ratio, p_val = stats.fisher_exact(table)
```
