```python
from scipy import stats

data1 = [12, 14, 15, 13, 16, 14, 15, 14]  # N = 8 (minimum required for normaltest)
data2 = [18, 20, 19, 21, 17, 22, 20, 19]

shapiro_stat, p_val = stats.shapiro(data1)
dagostino_stat, p_val = stats.normaltest(data1)
anderson_res = stats.anderson(data1, dist='norm')

levene_stat, p_val = stats.levene(data1, data2)
bartlett_stat, p_val = stats.bartlett(data1, data2)
```
