```python
import numpy as np
from scipy import stats

data = np.array([10, 12, 12, 13, 12, 11, 14, 13, 15, 100])

# Comprehensive summary (nobs, minmax, mean, variance, skewness, kurtosis)
summary = stats.describe(data)

# Individual robust & standard metrics
mode_res = stats.mode(data, keepdims=False)
iqr_val = stats.iqr(data)                # Interquartile range (Q3 - Q1)
sem_val = stats.sem(data)                # Standard error of the mean
trim_mean = stats.trim_mean(data, 0.1)   # Mean after trimming 10% from both ends
gmean_val = stats.gmean(data)            # Geometric mean
hmean_val = stats.hmean(data)            # Harmonic mean

# Transformations & Standardization
z_scores = stats.zscore(data)            # Standardize to mean=0, std=1
boxcox_data, best_lambda = stats.boxcox(data) # Power transform for normality (data > 0)
ranks = stats.rankdata(data)             # Assign ranks to data (handles ties)
```
