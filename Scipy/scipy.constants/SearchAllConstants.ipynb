```python
from scipy import constants

# Search for any constants matching a keyword string
matches = constants.find("gravitation")
print("Matching keys:", matches)
# Output: ['Newtonian constant of gravitation', 'Newtonian constant of gravitation over h-bar c']

# Retrieve the (value, unit, uncertainty) tuple for a specific key
val, unit, uncertainty = constants.physical_constants['Newtonian constant of gravitation']
print(f"Value: {val} {unit} (± {uncertainty})")
# Output: Value: 6.6743e-11 m^3 kg^-1 s^-2 (± 1.5e-15)

# Alternatively, use helper functions for specific metrics:
print("Value:", constants.value("Planck constant"))
print("Unit:", constants.unit("Planck constant"))
print("Relative Precision:", constants.precision("Planck constant"))
```
