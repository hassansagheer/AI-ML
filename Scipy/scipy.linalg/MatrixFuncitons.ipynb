```python
import numpy as np
from scipy import linalg

A = np.array([[1.0, 2.0], [0.0, 3.0]])

# Matrix exponential (e^A)
exp_A = linalg.expm(A)

# Matrix logarithm
log_A = linalg.logm(A)

# Matrix square root (sqrtm(A) @ sqrtm(A) = A)
sqrt_A = linalg.sqrtm(A)

# Matrix sine and cosine
sin_A = linalg.sinm(A)
cos_A = linalg.cosm(A)

print("Matrix Exponential:\n", exp_A)
print("Matrix Square Root:\n", sqrt_A)
print("Matrix Logarithm:\n", log_A)
print("Matrix sine:\n", sin_A)
print("Matrix cosine:\n", cos_A)
```
