```python
import numpy as np
from scipy import linalg

A = np.array([[3, 2, 0], [1, -1, 0], [0, 5, 1]])
b = np.array([2, 4, -1])

# General linear system
x = linalg.solve(A, b)

# Triangular system (faster when matrix is lower/upper triangular)
L = np.array([[3, 0, 0], [2, 1, 0], [1, 0, 1]])
x_tri = linalg.solve_triangular(L, b, lower=True)

# Least-squares solution (for overdetermined/underdetermined systems)
A_rect = np.array([[1, 1], [1, 2], [1, 3]])
b_rect = np.array([1, 2, 2])
x_lstsq, residuals, rank, s = linalg.lstsq(A_rect, b_rect)

print("General Solution:", x)
print("Triangular Solution:", x_tri)
print("Least Squares Solution:", x_lstsq)
```
