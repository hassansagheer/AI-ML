```python
import numpy as np
from scipy import linalg

A = np.array([[1, 2, 3], [4, 5, 6], [7, 8, 10]])

# LU Decomposition (P @ L @ U = A)
P, L, U = linalg.lu(A)

# QR Decomposition (Q @ R = A)
Q, R = linalg.qr(A)

# Singular Value Decomposition (U @ diag(s) @ Vh = A)
U_svd, s, Vh = linalg.svd(A)

# Cholesky Decomposition (for symmetric positive-definite matrices: L @ L.T = A)
SPD = np.array([[4, 2], [2, 3]])
L_chol = linalg.cholesky(SPD, lower=True)

# Schur Decomposition (Z @ T @ Z.H = A)
T, Z = linalg.schur(A)

print("Singular Values:", s)
print("Cholesky Factor:\n", L_chol)
```
