```python
import numpy as np
from scipy import linalg

A = np.array([[4, 3], [6, 3]])

# Determinant
det_A = linalg.det(A)

# Matrix Inverse
inv_A = linalg.inv(A)

# Matrix and Vector Norms
frobenius_norm = linalg.norm(A, 'fro')
l2_norm = linalg.norm(A, 2)

print("Determinant:", det_A)
print("Inverse:\n", inv_A)
print("Frobenius Norm:", frobenius_norm)
```
