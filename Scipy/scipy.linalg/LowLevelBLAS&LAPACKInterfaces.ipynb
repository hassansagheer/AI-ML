```python
import numpy as np
from scipy.linalg import blas, lapack

A = np.array([[1.0, 2.0], [3.0, 4.0]], order='F')
B = np.array([[5.0, 6.0], [7.0, 8.0]], order='F')

# Direct BLAS dgemm (Double-precision General Matrix Multiply: alpha * A @ B + beta * C)
C = blas.dgemm(alpha=1.0, a=A, b=B)

# Direct LAPACK dgetrf (LU factorization)
lu, piv, info = lapack.dgetrf(A)

print("BLAS DGEMM Result:\n", C)
print("LAPACK LU Info (0 = success):", info)
```
