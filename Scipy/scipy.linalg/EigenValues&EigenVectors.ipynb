```python
import numpy as np
from scipy import linalg

A = np.array([[1, -2], [1, 4]])

# Standard eigenvalues and right eigenvectors
eigvals, eigvecs = linalg.eig(A)

# Hermitian / Real Symmetric matrices (faster and guarantees real eigenvalues)
H = np.array([[2, 1], [1, 2]])
h_vals, h_vecs = linalg.eigh(H)

# Generalized eigenvalue problem: A @ v = w * B @ v
B = np.array([[2, 0], [0, 3]])
gen_vals, gen_vecs = linalg.eig(A, B)

print("Eigenvalues:", eigvals)
print("Hermitian Eigenvalues:", h_vals)
print("Generalized Eigenvalues:", gen_vals)
```
