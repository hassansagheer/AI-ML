```python
import numpy as np
from scipy import linalg

A = np.array([[-3, -2], [1, 0]])
B = np.array([[-1, 0], [0, -2]])
Q = np.array([[1, 0], [0, 1]])

# Sylvester equation: A @ X + X @ B = Q
X_sylv = linalg.solve_sylvester(A, B, Q)

# Continuous Lyapunov equation: A @ X + X @ A.H = Q
X_lyap = linalg.solve_continuous_lyapunov(A, Q)

# Continuous Algebraic Riccati Equation (CARE)
R = np.array([[1]])
B_ctrl = np.array([[0], [1]])
X_care = linalg.solve_continuous_are(A, B_ctrl, Q, R)

print("Sylvester Solution:\n", X_sylv)
print("Riccati Solution:\n", X_care)
```
