```python
from scipy.optimize import root_scalar

# Solve x^3 - x - 2 = 0 on bracket [1, 2]
def f(x):
    return x**3 - x - 2

sol = root_scalar(f, bracket=[1, 2], method='brentq')
print("Root:", sol.root)
```

```python
# Multivariate Root
import numpy as np
from scipy.optimize import root

# Solve system:
# x0 + 2*x1 - 2 = 0
# x0^2 + 4*x1^2 - 4 = 0
def system(x):
    return [
        x[0] + 2 * x[1] - 2,
        x[0]**2 + 4 * x[1]**2 - 4
    ]

sol = root(system, x0=[1.0, 1.0], method='hybr')
print("System roots:", sol.x)
```
