```python
from scipy.optimize import linprog

# Minimize: -x0 - 2*x1
c = [-1, -2]

# Subject to:
#  2*x0 + x1 <= 20
# -4*x0 + 5*x1 <= 10
A_ub = [
    [2, 1],
    [-4, 5]
]
b_ub = [20, 10]

# Bounds: x0 >= 0, x1 >= 0
bounds = [(0, None), (0, None)]

res = linprog(c, A_ub=A_ub, b_ub=b_ub, bounds=bounds, method='highs')
print("Optimal x:", res.x)
print("Optimal objective:", res.fun)
```

```python
import numpy as np
from scipy.optimize import milp, LinearConstraint, Bounds

# Maximize x0 + x1  ->  Minimize -x0 - x1
c = np.array([-1.0, -1.0])

# Constraint: 2*x0 + 3*x1 <= 12
A = np.array([[2.0, 3.0]])
constraints = LinearConstraint(A, -np.inf, 12.0)

# integrality: 1 = integer, 0 = continuous
integrality = np.array([1, 1])
bounds = Bounds([0, 0], [np.inf, np.inf])

res = milp(c=c, integrality=integrality, bounds=bounds, constraints=constraints)
print("Optimal integer x:", res.x)
```

```python
import numpy as np
from scipy.optimize import linear_sum_assignment

# Cost matrix (workers x tasks)
cost = np.array([
    [4, 1, 3],
    [2, 0, 5],
    [3, 2, 2]
])

row_ind, col_ind = linear_sum_assignment(cost)
print("Row indices:", row_ind)
print("Assigned columns:", col_ind)
print("Total minimum cost:", cost[row_ind, col_ind].sum())
```
