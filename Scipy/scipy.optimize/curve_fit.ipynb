```python
import numpy as np
from scipy.optimize import curve_fit

# Model: y = a * exp(-b * x) + c
def model(x, a, b, c):
    return a * np.exp(-b * x) + c

# Synthetic noisy data
x_data = np.linspace(0, 4, 50)
y_data = model(x_data, 2.5, 1.3, 0.5) + 0.1 * np.random.default_rng(42).normal(size=len(x_data))

popt, pcov = curve_fit(model, x_data, y_data, p0=[1.0, 1.0, 1.0])
print("Fitted parameters (a, b, c):", popt)
```

```python
# least_squares
import numpy as np
from scipy.optimize import least_squares

# Minimize sum of squares of residuals: r_i(x)
def residuals(x):
    return np.array([
        10 * (x[1] - x[0]**2),
        1 - x[0]
    ])

x0 = np.array([-1.2, 1.0])
res = least_squares(residuals, x0, bounds=([-2, -2], [2, 2]))

print("Optimal x:", res.x)
```
