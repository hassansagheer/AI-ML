```python
# unconstraint minimization (BFGS/Nelder-Mead)
import numpy as np
from scipy.optimize import minimize

# Rosenbrock function: minimum at (1, 1)
def objective(x):
    return (1 - x[0])**2 + 100 * (x[1] - x[0]**2)**2

x0 = np.array([-1.2, 1.0])
res = minimize(objective, x0, method='BFGS')

print("Optimal x:", res.x)
print("Function value:", res.fun)
```

```python
# constraint minimization with bounds (SLSQP)
import numpy as np
from scipy.optimize import minimize

# Minimize x^2 + y^2 subject to x + y >= 1 and 0 <= x, y <= 2
def objective(x):
    return x[0]**2 + x[1]**2

# Inequality constraint: fun(x) >= 0  ->  x[0] + x[1] - 1 >= 0
constraints = [
    {'type': 'ineq', 'fun': lambda x: x[0] + x[1] - 1}
]
bounds = [(0, 2), (0, 2)]

x0 = np.array([0.5, 0.5])
res = minimize(objective, x0, method='SLSQP', bounds=bounds, constraints=constraints)

print("Optimal x:", res.x)
```

```python
from scipy.optimize import minimize_scalar

# Minimize f(x) = (x - 2)^2 + 3
def f(x):
    return (x - 2)**2 + 3

# Bounded search between -5 and 5
res = minimize_scalar(f, bounds=(-5, 5), method='bounded')

print("Optimal x:", res.x)
print("Minimum value:", res.fun)
```

```python
import numpy as np
from scipy.optimize import differential_evolution

# Rastrigin function (highly multimodal, global min at 0)
def rastrigin(x):
    return 10 * len(x) + np.sum(x**2 - 10 * np.cos(2 * np.pi * x))

bounds = [(-5.12, 5.12), (-5.12, 5.12)]
res = differential_evolution(rastrigin, bounds, seed=42)

print("Global minimum x:", res.x)
print("Global minimum value:", res.fun)
```

```python
# Dual Annealing
import numpy as np
from scipy.optimize import dual_annealing

def rastrigin(x):
    return 10 * len(x) + np.sum(x**2 - 10 * np.cos(2 * np.pi * x))

bounds = [(-5.12, 5.12), (-5.12, 5.12)]
res = dual_annealing(rastrigin, bounds, seed=42)

print("Global minimum x:", res.x)
```
