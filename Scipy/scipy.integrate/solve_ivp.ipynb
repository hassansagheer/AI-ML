```python
import numpy as np
from scipy import integrate


# Example: Damped Harmonic Oscillator
# y'' + 0.5*y' + 4*y = 0  -->  y0' = y1,  y1' = -0.5*y1 - 4*y0
def oscillator(t, y):
    return [y[1], -0.5 * y[1] - 4.0 * y[0]]


# Event detection: stop or record when position y[0] crosses 0
def hit_zero(t, y):
    return y[0]


hit_zero.terminal = False  # Keep integrating after event
hit_zero.direction = -1  # Only trigger when crossing downward

t_span = (0, 10)
y0 = [1.0, 0.0]  # Initial position = 1, velocity = 0
t_eval = np.linspace(0, 10, 50)

sol = integrate.solve_ivp(
    oscillator,
    t_span,
    y0,
    method="RK45",  # Options: 'RK45', 'RK23', 'DOP853', 'Radau', 'BDF', 'LSODA'
    t_eval=t_eval,
    events=hit_zero,
    dense_output=True,  # Enables sol.sol(t) continuous interpolation
)

print(f"solve_ivp status: {sol.message}")
print(f"Zero-crossing times: {sol.t_events[0]}")
print(f"Interpolated value at t=2.5: {sol.sol(2.5)[0]:.6f}")
```
