```python
import numpy as np
from scipy import integrate

# --- 1D General Purpose Integration (quad) ---
# Integrates f(x) = exp(-x^2) from 0 to infinity
res_quad, err_quad = integrate.quad(lambda x: np.exp(-x**2), 0, np.inf)
print(f"quad: {res_quad:.6f} (err: {err_quad:.2e})")

# --- 1D Vectorized Integration (quad_vec) ---
# Integrates a vector-valued function f(x) = [x, x^2, x^3] from 0 to 1
res_vec, err_vec = integrate.quad_vec(lambda x: np.array([x, x**2, x**3]), 0, 1)
print(f"quad_vec: {res_vec}")

# --- 2D Integration (dblquad) ---
# Integrates f(y, x) = x * y^2 over x in [0, 2] and y in [0, 1]
# Note: integrand arguments must be ordered (y, x)
res_dbl, _ = integrate.dblquad(lambda y, x: x * y**2, 0, 2, 0, 1)
print(f"dblquad: {res_dbl:.6f}")

# --- 3D Integration (tplquad) ---
# Integrates f(z, y, x) = x * y * z over unit cube [0, 1]^3
# Note: integrand arguments must be ordered (z, y, x)
res_tpl, _ = integrate.tplquad(lambda z, y, x: x * y * z, 0, 1, 0, 1, 0, 1)
print(f"tplquad: {res_tpl:.6f}")

# --- N-Dimensional Integration (nquad) ---
# Integrates f(x0, x1) = x0 * x1 over [0, 1] x [0, 2]
res_nquad, _ = integrate.nquad(lambda x0, x1: x0 * x1, [[0, 1], [0, 2]])
print(f"nquad: {res_nquad:.6f}")

# --- Fixed-Order Gaussian Quadrature (fixed_quad) ---
# Fast integration using n=5 Gauss-Legendre points (no error estimate)
res_fixed, _ = integrate.fixed_quad(lambda x: np.sin(x), 0, np.pi, n=5)
print(f"fixed_quad: {res_fixed:.6f}")

# --- Adaptive Gaussian Quadrature (quadrature) ---
# Increases order until tolerance is met
res_gauss, err_gauss = integrate.quad(lambda x: np.sin(x), 0, np.pi)
print(f"quad: {res_gauss:.6f}")

# --- Adaptive Tanh-Sinh Quadrature ---
res_tanhsinh = integrate.tanhsinh(lambda x: np.sin(x), 0, np.pi)
print(f"tanhsinh: {res_tanhsinh.integral:.6f}")
```
