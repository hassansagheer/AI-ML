```python
import numpy as np
from scipy import integrate

# Define the time points for odeint
t_eval = np.linspace(0.0, 5.0, 100)


# --- Legacy ODEINT (Wraps Fortran LSODA) ---
# Signature: f(y, t)
def oscillator_legacy(y, t):
  return [y[1], -0.5 * y[1] - 4.0 * y[0]]


y_odeint = integrate.odeint(oscillator_legacy, y0=[1.0, 0.0], t=t_eval)
print(f"odeint final position: {y_odeint[-1, 0]:.6f}")


# --- Manual Step-by-Step Solver (RK45 / BDF / Radau / DOP853 / LSODA) ---
# Signature: f(t, y) — argument order is reversed compared to odeint
def oscillator(t, y):
  return [y[1], -0.5 * y[1] - 4.0 * y[0]]


solver = integrate.RK45(oscillator, t0=0.0, y0=[1.0, 0.0], t_bound=5.0)
while solver.status == "running":
  solver.step()
print(f"Manual RK45 reached t={solver.t:.2f}, y={solver.y[0]:.6f}")
```
