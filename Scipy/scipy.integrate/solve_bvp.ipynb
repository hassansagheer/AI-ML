```python
import numpy as np
from scipy import integrate


# Solve Bratu's problem: y'' + exp(y) = 0, with y(0) = 0 and y(1) = 0
# Convert to 1st-order system: y0' = y1,  y1' = -exp(y0)
def bvp_ode(x, y):
    return np.vstack((y[1], -np.exp(y[0])))


# Boundary condition residuals (must equal 0 at solution)
def bvp_bc(ya, yb):
    return np.array([ya[0], yb[0]])  # ya[0] - 0 = 0, yb[0] - 0 = 0


x_mesh = np.linspace(0, 1, 10)
# Initial guess for y (shape: [n_equations, n_mesh_points])
y_guess = np.zeros((2, x_mesh.size))

bvp_res = integrate.solve_bvp(bvp_ode, bvp_bc, x_mesh, y_guess)
print(f"solve_bvp converged: {bvp_res.success}")
print(f"Peak value at x=0.5: {bvp_res.sol(0.5)[0]:.6f}")
```
