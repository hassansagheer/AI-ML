```python
import numpy as np
from scipy import integrate

x = np.linspace(0, np.pi, 11)
y = np.sin(x)

# --- Trapezoidal Rule (trapezoid) ---
# Note: Replaced the deprecated `trapz` in SciPy 1.12+
area_trap = integrate.trapezoid(y, x)
print(f"trapezoid: {area_trap:.6f}")

# --- Cumulative Trapezoidal Rule (cumulative_trapezoid) ---
# Returns an array of the running integral (useful for velocity -> position)
cum_trap = integrate.cumulative_trapezoid(y, x, initial=0)
print(f"cumulative_trapezoid (final val): {cum_trap[-1]:.6f}")

# --- Simpson's Rule (simpson) ---
# Fits piecewise quadratics; generally much more accurate than trapezoid
area_simp = integrate.simpson(y, x=x)
print(f"simpson: {area_simp:.6f}")

# --- Cumulative Simpson's Rule (cumulative_simpson) ---
cum_simp = integrate.cumulative_simpson(y, x=x, initial=0)
print(f"cumulative_simpson (final val): {cum_simp[-1]:.6f}")

# --- Romberg Integration for Samples (romb) ---
# Requires 2^k + 1 evenly spaced samples
x_romb = np.linspace(0, np.pi, 17)  # 2^4 + 1 = 17 points
y_romb = np.sin(x_romb)
dx = x_romb[1] - x_romb[0]
area_romb_samples = integrate.romb(y_romb, dx=dx)
print(f"romb (samples): {area_romb_samples:.6f}")
```
