```python
import numpy as np
from scipy.interpolate import BSpline, make_interp_spline

x = np.linspace(0, 10, 15)
y = np.sin(x)
x_new = np.linspace(0, 10, 100)

# Create an interpolating B-spline of degree k=3 (cubic)
bspl = make_interp_spline(x, y, k=3)
y_new = bspl(x_new)

# Or construct directly from knots (t), coefficients (c), and degree (k)
t, c, k = bspl.t, bspl.c, bspl.k
custom_bspl = BSpline(t, c, k)
```
