```python
import numpy as np
from scipy.interpolate import Akima1DInterpolator, PchipInterpolator

x = np.array([0, 1, 2, 3, 4, 5, 6], dtype=float)
y = np.array([0, 0, 0, 1, 1, 1, 0], dtype=float)
x_new = np.linspace(0, 6, 200)

# PCHIP (Piecewise Cubic Hermite Interpolating Polynomial) - preserves monotonicity
pchip = PchipInterpolator(x, y)
y_pchip = pchip(x_new)

# Akima - smoother visual curve, local to 5 neighboring points
akima = Akima1DInterpolator(x, y)
y_akima = akima(x_new)
```
