```python
import numpy as np
from scipy.interpolate import griddata

rng = np.random.default_rng(0)
points = rng.uniform(0, 1, (100, 2))  # 100 scattered (x, y) points
values = np.sin(points[:, 0] * 5) * np.cos(points[:, 1] * 5)

# Target grid
grid_x, grid_y = np.mgrid[0:1:50j, 0:1:50j]

# method options: 'nearest', 'linear', 'cubic' (cubic is 2D only)
grid_z = griddata(points, values, (grid_x, grid_y), method="cubic")
print(grid_z)
```
