```python
import numpy as np
from scipy.interpolate import SmoothBivariateSpline

rng = np.random.default_rng(0)
x = rng.uniform(-2, 2, 200)
y = rng.uniform(-2, 2, 200)
z = np.exp(-(x**2 + y**2)) + rng.normal(0, 0.05, 200)

# s controls smoothing; bbox defines the domain boundaries
sbs = SmoothBivariateSpline(x, y, z, s=0.5)

x_grid = np.linspace(-2, 2, 50)
y_grid = np.linspace(-2, 2, 50)
Z_smooth = sbs(x_grid, y_grid)
```
