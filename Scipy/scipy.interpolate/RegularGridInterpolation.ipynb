```python
import numpy as np
from scipy.interpolate import RegularGridInterpolator

x = np.linspace(0, 5, 10)
y = np.linspace(0, 5, 15)
X, Y = np.meshgrid(x, y, indexing="ij")
Z = np.sin(X) * np.cos(Y)

# method options: 'linear', 'nearest', 'slinear', 'cubic', 'quintic', 'pchip'
interp = RegularGridInterpolator((x, y), Z, method="cubic")

# Query at specific (x, y) coordinates
pts = np.array([[1.2, 3.4], [4.5, 0.8]])
values = interp(pts)
print("RegularGridInterpolation N-D: ", values)
```

```python
# 2D Bivariate Spline
import numpy as np
from scipy.interpolate import RectBivariateSpline

x = np.linspace(0, 4, 20)
y = np.linspace(0, 4, 20)
X, Y = np.meshgrid(x, y, indexing="ij")
Z = np.exp(-(X - 2) ** 2 - (Y - 2) ** 2)

# kx, ky specify spline degrees (default 3); s=0 for exact interpolation
spline_2d = RectBivariateSpline(x, y, Z, kx=3, ky=3, s=0)

x_new = np.linspace(0, 4, 100)
y_new = np.linspace(0, 4, 100)
Z_new = spline_2d(x_new, y_new)  # Returns a 100x100 grid
print("RectBivariateSpline: ", Z_new)
```
