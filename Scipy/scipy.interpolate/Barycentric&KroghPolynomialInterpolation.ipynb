```python
import numpy as np
from scipy.interpolate import BarycentricInterpolator, KroghInterpolator

x = np.linspace(0, 5, 6)
y = np.exp(-x)
x_new = np.linspace(0, 5, 100)

# Barycentric (numerically stable for adding points dynamically)
bary = BarycentricInterpolator(x, y)
y_bary = bary(x_new)

# Krogh (also supports specifying derivatives at points)
krogh = KroghInterpolator(x, y)
y_krogh = krogh(x_new)
```
