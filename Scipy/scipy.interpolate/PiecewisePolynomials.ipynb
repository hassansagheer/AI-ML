```python
import numpy as np
from scipy.interpolate import BPoly, PPoly

# Breakpoints
x = np.array([0.0, 1.0, 2.0])

# Coefficients in power basis: c[m, i] is coeff of (x - x[i])**(k - m)
# Interval 0: 1*x^2 + 0*x + 0
# Interval 1: -1*(x-1)^2 + 2*(x-1) + 1
c = np.array([[1.0, -1.0], [0.0, 2.0], [0.0, 1.0]])

pp = PPoly(c, x)
print("Value at 1.5:", pp(1.5))
print("Roots:", pp.roots())

# Antiderivative and derivative
pp_deriv = pp.derivative()
pp_integ = pp.antiderivative()

# Bernstein polynomial basis (BPoly) using function values and derivatives at endpoints
bp = BPoly.from_derivatives([0, 1], [[0, 1], [1, -1]])  # f(0)=0, f'(0)=1; f(1)=1, f'(1)=-1
print("BPoly at 0.5:", bp(0.5))
```
