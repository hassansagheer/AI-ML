```python
import numpy as np
from scipy.interpolate import (
    CloughTocher2DInterpolator,
    LinearNDInterpolator,
    NearestNDInterpolator,
)

rng = np.random.default_rng(0)
points = rng.uniform(0, 1, (100, 2))
values = points[:, 0] ** 2 + points[:, 1] ** 2

# 1. Nearest neighbor (N-D)
nn_interp = NearestNDInterpolator(points, values)

# 2. Piecewise linear (N-D)
lin_interp = LinearNDInterpolator(points, values)

# 3. Smooth C1 piecewise cubic (2D only)
ct_interp = CloughTocher2DInterpolator(points, values)

query_pts = np.array([[0.25, 0.25], [0.75, 0.5]])
res = ct_interp(query_pts)
```
