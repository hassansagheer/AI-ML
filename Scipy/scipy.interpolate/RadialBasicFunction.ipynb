```python
import numpy as np
from scipy.interpolate import RBFInterpolator

rng = np.random.default_rng(0)
points = rng.uniform(-1, 1, (50, 3))  # 50 scattered 3D points
values = np.sum(points**2, axis=1)

# kernel options: 'thin_plate_spline' (default), 'multiquadric', 'inverse_multiquadric',
#                 'gaussian', 'linear', 'cubic', 'quintic'
rbf = RBFInterpolator(points, values, kernel="thin_plate_spline", smoothing=0.0)

query_pts = np.array([[0.0, 0.0, 0.0], [0.5, -0.5, 0.2]])
pred = rbf(query_pts)
```
