```python
import numpy as np
from scipy.interpolate import interp1d

x = np.linspace(0, 10, 11)
y = np.sin(x)
x_new = np.linspace(0, 10, 100)

# Recommended for 1D linear
y_linear_np = np.interp(x_new, x, y)

# Legacy interp1d supporting multiple modes ('linear', 'nearest', 'zero', 'slinear', 'quadratic', 'cubic')
f_nearest = interp1d(x, y, kind="nearest")
y_nearest = f_nearest(x_new)
```
